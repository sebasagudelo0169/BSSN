# **Se desarrolla el codigo para resolver el caso 1+1 del formalismo BSSN**


Aunque en un inicion parece un caso, muy trivial ya que en esta dimension no hay ninguna propiedad fisica, se usa para estudiar la evolucion del sistema segun los gauges escogidos

**Variables del sistema**

En la formulación BSSN estándar tenemos:$$\gamma_{ij} = e^{4\phi}\tilde{\gamma}_{ij} \quad \text{o equivalentemente} \quad \psi = e^{4\phi}$$Donde las variables fundamentales completas son:$$\{\phi, \tilde{\gamma}_{ij}, K, \tilde{A}_{ij}, \tilde{\Gamma^i}, \alpha\}$$Sin embargo, en un espacio-tiempo de $1+1$ dimensiones
*   Curvatura Extrínseca sin Traza ($\tilde{A}_{ij}$): En 1D, se define restando la traza: $A_{ij} = K_{ij} - \frac{1}{1}\gamma_{ij}K$. En 1D, la traza es exactamente $K = \gamma^{xx}K_{xx} \implies K_{xx} = \gamma K$. Si restamos la traza para encontrar la parte sin traza en 1D:$$A_{xx} = K_{xx} - \gamma_{xx}K = K_{xx} - K_{xx} = 0$$Por lo tanto, la parte conforme sin traza es idénticamente cero: $\tilde{A}_{xx} = 0$.
*   Funciones de Conexión Conforme ($\tilde{\Gamma^i}$): Dado quela metrica conforme es una constante espacial, los símbolos de Christoffel conformes son cero. $\tilde{\Gamma^x} = 0 $


Como puedes ver, todas las variables tensoriales conformes de BSSN ($\tilde{\gamma}_{ij}$, $\tilde{A}_{ij}$, $\tilde{\Gamma^i}$) "desaparecen" en 1+1. Las únicas variables dinámicas que sobreviven son los escalares: el factor conforme $\phi$ y la traza de la curvatura extrínseca $K$.
Al reducir las ecuaciones originales de BSSN y sustituir nuestros valores triviales ($\tilde{A}_{xx}=0, \tilde{\Gamma^x}=0$), las ecuaciones de evolución se reducen dramáticamente a solo dos, La ecuacionde evolucion de $\phi$ para darle forma a nuestro sistema resultante de un ley de conservacion sera sustituida por la ecuacion de evolucion de gamma en AMD .

por tanto, nuestras variables son :$$\{\gamma_{xx}, K, \alpha\}$$La estrategia de evolución para este esquema híbrido es la siguiente:

*   $\gamma_{xx}$ evoluciona mediante ADM.
*   $K$ evoluciona mediante BSSN/ADM.
*   $\alpha$ evoluciona mediante la condición de Bona-Massó.



**Métrica espacial**

En una única dimensión espacial, el intervalo métrico se reduce a:$$dl^2 = \gamma_{xx}(t,x) \, dx^2$$Por lo tanto, la única componente no trivial es $\gamma \equiv \gamma_{xx}$, y su inversa viene dada simplemente por:$$\gamma^{xx} = \frac{1}{\gamma}$$
Ecuación ADM para la métricaLa ecuación general de evolución de ADM para la métrica espacial es:$$\partial_t \gamma_{ij} = -2\alpha K_{ij} + \mathcal{L}_\beta \gamma_{ij}$$Al no considerar vector de desplazamiento ($\beta = 0$), el término de la derivada de Lie se hace nulo desaparece ($\mathcal{L}_\beta \gamma_{ij} = 0$), quedando:$$\partial_t \gamma_{ij} = -2\alpha K_{ij}$$En una dimensión (1D), la curvatura extrínseca se descompone en su parte traza y su parte sin traza ($A_{xx}$):$$K_{xx} = \frac{1}{3}\,\gamma_{xx}K + A_{xx}$$. En una dimensión el único componente coincide plenamente con su traza:$$K_{xx} = \gamma_{xx}K$$Sustituyendo esto en la evolución de la métrica, obtenemos nuestra primera ecuación del sistema:$$\partial_t \gamma = -2\alpha\gamma K$$

**Relación con el factor conforme**

Para comprobar la consistencia, recordemos que BSSN define el factor conforme $\phi$ como:$$\phi = \frac{1}{12}\ln(\det\gamma_{ij})$$En nuestro caso 1D, $\det\gamma_{ij} = \gamma$, por lo que $\phi = \frac{1}{12}\ln\gamma$. Si calculamos su derivada temporal obtenemos:$$\partial_t \phi = \frac{1}{12} \frac{1}{\gamma} \partial_t \gamma$$Sustituyendo la ecuación ADM previamente calculada ($\partial_t \gamma = -2\alpha\gamma K$):$$\partial_t \phi = -\frac{1}{6}\alpha K$$Esta es exactamente la ecuación estándar de evolución de BSSN para $\phi$, lo que demuestra que sustituir el factor conforme directamente por la métrica $\gamma$ en 1D es formalmente consistente

**Evolución de la curvatura extrínseca**

La ecuación de evolución de BSSN para la traza de la curvatura extrínseca $K$ en el vacío es:$$\partial_t K = -D^iD_i\alpha + \alpha \left( \tilde{A}_{ij}\tilde{A}^{ij} + \frac{1}{3}K^2 \right)$$ sabemos que ($\tilde{A}_{ij} = 0$), la expresión se reduce a:$$\partial_t K = -D^iD_i\alpha + \frac{1}{3}\alpha K^2$$
**Laplaciano covariante en 1D:**
El laplaciano del lapso se puede desarrollar usando la inversa de la métrica y la derivada covariante:$$D^iD_i\alpha = \gamma^{xx}D_xD_x\alpha = \frac{1}{\gamma} \left( \partial_x^2\alpha - \Gamma^x_{xx}\partial_x\alpha \right)$$Dado que el único símbolo de Christoffel no nulo en este escenario es:$$\Gamma^x_{xx} = \frac{1}{2\gamma}\partial_x\gamma$$El laplaciano covariante toma la forma final:$$D^iD_i\alpha = \frac{1}{\gamma} \left( \alpha_{xx} - \frac{\gamma_x}{2\gamma}\alpha_x \right)$$Donde se denota de forma compacta $\alpha_{xx} \equiv \partial_x^2\alpha$, $\alpha_x \equiv \partial_x\alpha$ y $\gamma_x \equiv \partial_x\gamma$. Sustituyendo este resultado, obtenemos la ecuación de evolución para $K$:$$\partial_t K = -\frac{1}{\gamma} \left( \alpha_{xx} - \frac{\gamma_x}{2\gamma}\alpha_x \right) + \frac{1}{3}\alpha K^2$$

**Gauge Bona-Massó**

Para la función de lapso $\alpha$, adoptamos la familia genérica de condiciones de rebanado dinámico de Bona-Massó:$$\partial_t\alpha = -\alpha^2 f(\alpha) K$$Dependiendo de la elección de la función $f(\alpha)$, podemos recuperar condiciones de gauge muy conocidas en relatividad numérica:

*   Harmonic slicing: Si $f(\alpha)=1$ luego $\partial_t\alpha = -\alpha^2 K $
*   Condición  1+log:  Si $f(\alpha) = \frac{2}{\alpha}$luego $\partial_t\alpha = -2\alpha K$




**Sistema final híbrido ADM-BSSN**

Uniendo las piezas, el conjunto de variables dinámicas incógnitas es $\{\gamma(x,t), K(x,t), \alpha(x,t)\}$, cuyas evoluciones quedan gobernadas por el siguiente sistema de ecuaciones en derivadas parciales:$$\partial_t \gamma = -2\alpha\gamma K$$$$\partial_t K = -\frac{1}{\gamma} \left( \alpha_{xx} - \frac{\gamma_x}{2\gamma}\alpha_x \right) + \frac{1}{3}\alpha K^2$$$$\partial_t\alpha = -\alpha^2 f(\alpha) K$$


Partiendo de las ecuaciones de evolución anteriores, es posible aplicar un cambio de variables estratégico para simplificar la estructura del problema. En primer lugar, introducimos las variables auxiliares $D_\alpha$ y $D_\gamma$, las cuales se definen a partir de las derivadas espaciales de los logaritmos naturales de las funciones originales y de la componente métrica $\gamma$:
\begin{equation}
\gamma := \gamma_{xx}, \quad D_\alpha := \partial_x \ln \alpha \quad \text{and} \quad D_\gamma := \partial_x \ln \gamma.
\end{equation}

Al sustituir estas definiciones y aplicar el álgebra correspondiente, el sistema original se transforma en un nuevo conjunto de ecuaciones para las variables de primer orden:
\begin{align}
\partial_t D_\alpha + \partial_x (\alpha f K) &= 0 \ , \\
\partial_t D_\gamma + \partial_x (2\alpha K) &= 0 \ , \\
\partial_t K + \partial_x (\alpha D_\alpha / \gamma) &= \alpha \left( K^2 - D_\alpha D_\gamma / 2\gamma \right) \ .
\end{align}

A partir de este punto, se observa que la ecuación de evolución para la curvatura externa $K$ puede reescribirse elegantemente como una ley de conservación utilizando la variable escalada $g^{1/2}$:
\begin{equation}
\partial_t \left( \gamma^{1/2} K \right) + \partial_x \left( \alpha D_\alpha / \gamma^{1/2} \right) = 0 \ .
\end{equation}

Si ahora agrupamos nuestras variables en un vector de estado definido como $\vec{v} := (D_\alpha, D_\gamma, \tilde{K})$, donde se introduce la cantidad $\tilde{K} := g^{1/2} K$, las ecuaciones de evolución para las variables de primer orden pueden acoplarse y expresarse como un sistema conservativo de la forma:
\begin{equation}
\partial_t \vec{v} + \partial_x (\mathbf{M} \vec{v}) = 0 \ ,
\end{equation}
donde la matriz característica $\mathbf{M}$ queda completamente determinada por los coeficientes del sistema y viene dada por:
\begin{equation}
\mathbf{M} = \begin{pmatrix}
0 & 0 & \alpha f / g^{1/2} \\
0 & 0 & 2 \alpha / g^{1/2} \\
\alpha / g^{1/2} & 0 & 0
\end{pmatrix} \ .
\end{equation}

Esta sistema matricial es el que vamos a resolver

inicialmente supongo una metrica de minkowski y supongo una pequeña pertubarcion modelado por h.

\begin{align*}
\gamma &= 1 - h'^2 \implies D_\gamma = -2h'h'' / \gamma \ , \\
K &:= K^x_x = -h'' / \gamma^{3/2} \implies \tilde{K} = -h'' / \gamma \ .
\end{align*}

A la funcion h se le dara un perfil gaussiano
\begin{equation}
h = A \exp \left[ (x - x_c)^2 / \sigma^2 \right] \ .
\end{equation}

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
```

```python
# Constantes
A=5
sigma=10
f=1.69   # se tomara
tfinal= 75      # tiempo final de mi simulacion

x= sp.symbols('x')
h= A*sp.exp(-((x-150)**2)/(sigma**2)) # defino h como gaussiana centrada en 150 el punto medio de mi espacio
alpha = 1

# Calculo las derivadas de la funcion h en este caso gausiana de manera simbolica
dh_dx= sp.diff(h,x)
d2h_dx2= sp.diff(dh_dx,x)
dalpha_dx= sp.diff(alpha,x)

# Aqui las convierto a numerico
dh_numerica = sp.lambdify((x), h, 'numpy')
dh_dx_numerica = sp.lambdify((x), dh_dx, 'numpy')
d2h_dx2_numerica = sp.lambdify((x), d2h_dx2, 'numpy')
dalpha_dx_numerica = sp.lambdify((x), dalpha_dx, 'numpy')

# --- MALLA ESPACIAL ---------------------------------------------
N = 2500
x_numerico = np.linspace(0, 300, N)
dx = x_numerico[1] - x_numerico[0]

# Evaluaciones iniciales basadas en la métrica
dh_dx_evaluada = dh_dx_numerica(x_numerico)
d2h_dx2_evaluada = d2h_dx2_numerica(x_numerico)
Dalpha_init = dalpha_dx_numerica(x_numerico)             # Como no es un array me genera problemas

# Defino las condiciones inciales antes expuestas
g_init = 1 - dh_dx_evaluada**2
Dg_init = -2 * dh_dx_evaluada * d2h_dx2_evaluada / g_init
k_init = -d2h_dx2_evaluada / g_init
Dalpha_init = np.zeros_like(x_numerico)    # Corregido: Ahora es un vector de ceros
```

```python
def KO(u, dx, eps=0.9):
    """
    método de Disipación de Kreiss-Oliger de 4to orden
    se usa para eliminar el ruido en la solucion
    """
    q = np.zeros_like(u)

    q[2:-2] = -eps*(u[:-4]- 4*u[1:-3]+ 6*u[2:-2]- 4*u[3:-1]+ u[4:])/(16*dx)
    return q


def integrar_RK4(D, dx, initial):
    """
    Integra

        dy/dx = D(x)*y

    mediante Runge-Kutta de cuarto orden.
    """
    N = len(D)
    y = np.zeros(N)
    y[0] = initial

    for i in range(N-1):
        D0 = D[i]
        D1 = 0.5*(D[i] + D[i+1])
        D3 = D[i+1]

        k1 = dx*D0*y[i]
        k2 = dx*D1*(y[i] + 0.5*k1)
        k3 = dx*D2*(y[i] + 0.5*k2)
        k4 = dx*D3*(y[i] + k3)

        y[i+1] = y[i] + (k1 + 2*k2 + 2*k3 + k4)/6

    return y

def sistema_edp(t, y_global, N, dx, f):
    # 1. Desempaquetar variables
    Dalpha = y_global[0:N]
    Dg     = y_global[N:2*N]
    k      = y_global[2*N:3*N]

    # 2. Reconstrucción espacial de alpha y g
    alpha_frontera = 1.0
    g_frontera = 1.0

    alpha = integrar_RK4(Dalpha, dx=dx, initial=alpha_frontera)
    g     = integrar_RK4(Dg, dx=dx, initial=g_frontera)

    sqrt_g = np.sqrt(g)

    # 3. Flujos
    F = np.zeros((3, N))
    F[0, :] = (alpha * f / sqrt_g) * k
    F[1, :] = (2.0 * alpha / sqrt_g) * k
    F[2, :] = (alpha / sqrt_g) * Dalpha

    # 4. Derivada espacial del flujo (Diferencias finitas 4to orden)
    dF_dx = np.zeros_like(F)

    dF_dx[:,2:-2] = (-F[:,4:] + 8*F[:,3:-1] - 8*F[:,1:-3] + F[:,:-4])/(12*dx)

    # Fronteras
    dF_dx[:,0]  = (-25*F[:,0] + 48*F[:,1] - 36*F[:,2] + 16*F[:,3] - 3*F[:,4])/(12*dx)
    dF_dx[:,1]  = (-3*F[:,0] - 10*F[:,1] + 18*F[:,2] - 6*F[:,3] + F[:,4])/(12*dx)
    dF_dx[:,-2] = (3*F[:,-1] + 10*F[:,-2] - 18*F[:,-3] + 6*F[:,-4] - F[:,-5])/(12*dx)
    dF_dx[:,-1] = (25*F[:,-1] - 48*F[:,-2] + 36*F[:,-3] - 16*F[:,-4] + 3*F[:,-5])/(12*dx)

    # 5. Sistema evolutivo
    dDalpha_dt = -dF_dx[0]
    dDg_dt     = -dF_dx[1]
    dk_dt      = -dF_dx[2]

    # 6. Disipación KO a las variables de estado
    dDalpha_dt += KO(Dalpha, dx)
    dDg_dt     += KO(Dg, dx)
    dk_dt      += KO(k, dx)

    return np.concatenate([dDalpha_dt, dDg_dt, dk_dt])
```

```python
y_inicial = np.concatenate([Dalpha_init, Dg_init, k_init])     # concateno las condiciones iniciales

t_span = (0.0, tfinal)
t_eval = np.linspace(0, tfinal, 500)

# Resolver usando solve_ivp
solucion = solve_ivp(
    sistema_edp,
    t_span,
    y_inicial,
    t_eval=t_eval,
    args=(N, dx, f),
    method='RK45'
)
```

La solución obtenida corresponde a una matriz bidimensional. En ella, las filas contienen las variables del sistema: desde la fila 1 hasta la N se encuentran los valores de $D_α$	, desde la fila N hasta la 2N los de $D_g$
, y desde la fila 2N hasta la 3N los de K. Las columnas representan los distintos instantes temporales considerados, en este caso 50, distribuidos en el intervalo de 0 a 100.

```python
k_final = solucion.y[2*N:3*N,-1 ]  #  el -1 es para seleccionar del array solo el del ultimo intervalo de tiempo

plt.figure(figsize=(8, 5))
plt.plot(x_numerico, k_init, label='k Inicial (t=0)')
plt.plot(x_numerico, k_final, label=f'k Final (t={tfinal})')
plt.title('Evolución Temporal de la curvatura k')
plt.xlabel('Espacio (x)')
plt.ylabel('k')
plt.legend()
plt.grid(True)


Dalpha_t0 = solucion.y[0:N, 0]
Dalpha_t100 = solucion.y[0:N, -1]

# Definir la condición de frontera en x=0
alpha_frontera = 1.0  # α(0) = 1

# 3. Calcular la reconstrucción espacial de alpha

alpha_t0 = integrar_RK4(Dalpha_t0, dx, alpha_frontera)
alpha_t100 = integrar_RK4(Dalpha_t100, dx, alpha_frontera)

plt.figure(figsize=(8, 5))

plt.plot(x_numerico, alpha_t0, color='blue', label='$\\alpha$ en t = 0')
plt.plot(x_numerico, alpha_t100, color='red', label=f'$\\alpha$ en t = {tfinal}')

plt.title(f'Evolución del Lapse $\\alpha$ (t=0 vs t={tfinal})')
plt.xlabel('Espacio (x)')
plt.ylabel('$\\alpha(x)$')
plt.legend()
plt.grid(True)
plt.show()
```

```python
Dgamma_final = solucion.y[N:2*N,-1 ]  #  el -1 es para seleccionar del array solo el del ultimo intervalo de tiempo
Dgamma_init = solucion.y[N:2*N,0 ]  #  el -1 es para seleccionar del array solo el del ultimo intervalo de tiempo
g_frontera=1

gamma_final= integrar_RK4(Dgamma_final, dx, g_frontera)
gamma_init = integrar_RK4(Dgamma_init, dx, g_frontera)

plt.figure(figsize=(8, 5))
# plt.plot(x_numerico, g_init, label='g inicial  (t=0)')
plt.plot(x_numerico, gamma_init, label='gamma Inicial (t=0)')
plt.plot(x_numerico, gamma_final, label=f'gamma Final (t={tfinal})')
plt.title('Evolución Temporal de la metrical espacial')
plt.xlabel('Espacio (x)')
plt.ylabel('K')
plt.legend()
plt.grid(True)
```

Se implementó una condición inicial con perfil gaussiano para modelar una perturbación localizada y analizar su evolución dinámica. Al restringir el análisis a una simetría espacial de 1+1 dimensiones, el contenido de radiación física gravitacional es nulo; por lo tanto, la dinámica observada corresponde puramente a la propagación de los modos de gauge (coordenadas).Como se aprecia en el perfil de la curvatura extrínseca $K$, la perturbación se divide en dos frentes de onda que se propagan simétricamente en direcciones opuestas. No obstante, el esquema numérico exhibe una inestabilidad severa en los frentes o fronteras de la onda, donde la solución experimenta una divergencia
