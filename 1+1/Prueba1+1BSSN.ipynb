# **Se desarrolla el codigo para resolver el caso 1+1 del formalismo BSSN**

**Variables del sistema**

En la formulación BSSN estándar tenemos:$$\gamma_{ij} = e^{4\phi}\tilde{\gamma}_{ij} \quad \text{o equivalentemente} \quad \psi = e^{4\phi}$$Donde las variables fundamentales completas son:$$\{\phi, \tilde{\gamma}_{ij}, K, \tilde{A}_{ij}, \tilde{\Gamma^i}, \alpha\}$$Sin embargo, en un espacio-tiempo de $1+1$ dimensiones
*   Curvatura Extrínseca sin Traza ($\tilde{A}_{ij}$): En 1D, se define restando la traza: $A_{ij} = K_{ij} - \frac{1}{1}\gamma_{ij}K$. En 1D, la traza es exactamente $K = \gamma^{xx}K_{xx} \implies K_{xx} = \gamma K$. Si restamos la traza para encontrar la parte sin traza en 1D:$$A_{xx} = K_{xx} - \gamma_{xx}K = K_{xx} - K_{xx} = 0$$Por lo tanto, la parte conforme sin traza es idénticamente cero: $\tilde{A}_{xx} = 0$.
*   Funciones de Conexión Conforme ($\tilde{\Gamma^i}$): Dado quela metrica conforme es una constante espacial, los símbolos de Christoffel conformes son cero. $\tilde{\Gamma^x} = 0 $


Como puedes ver, todas las variables tensoriales conformes de BSSN ($\tilde{\gamma}_{ij}$, $\tilde{A}_{ij}$, $\tilde{\Gamma^i}$) "desaparecen" en 1+1. Las únicas variables dinámicas que sobreviven son los escalares: el factor conforme $\phi$ y la traza de la curvatura extrínseca $K$.
Al reducir las ecuaciones originales de BSSN y sustituir nuestros valores triviales ($\tilde{A}_{xx}=0, \tilde{\Gamma^x}=0$), las ecuaciones de evolución se reducen dramáticamente a solo dos, La ecuacionde evolucion de $\phi$ para darle forma a nuestro sistema resultante de un ley de conservacion sera sustituida por la ecuacion de evolucion de gamma en AMD .

por tanto, nuestras variables son :$$\{\gamma_{xx}, K, \alpha\}$$La estrategia de evolución para este esquema híbrido es la siguiente:

*   $\gamma_{xx}$ evoluciona mediante ADM.
*   $K$ evoluciona mediante BSSN/ADM.
*   $\alpha$ evoluciona mediante la condición de Bona-Massó.



**Relación con el factor conforme**

Para comprobar la consistencia, recordemos que BSSN define el factor conforme $\phi$ como:$$\phi = \frac{1}{12}\ln(\det\gamma_{ij})$$En nuestro caso 1D, $\det\gamma_{ij} = \gamma$, por lo que $\phi = \frac{1}{12}\ln\gamma$. Si calculamos su derivada temporal obtenemos:$$\partial_t \phi = \frac{1}{12} \frac{1}{\gamma} \partial_t \gamma$$Sustituyendo la ecuación ADM previamente calculada ($\partial_t \gamma = -2\alpha\gamma K$):$$\partial_t \phi = -\frac{1}{6}\alpha K$$Esta es exactamente la ecuación estándar de evolución de BSSN para $\phi$, lo que demuestra que sustituir el factor conforme directamente por la métrica $\gamma$ en 1D es formalmente consistente

**Sistema final híbrido ADM-BSSN**

Uniendo las piezas, el conjunto de variables dinámicas incógnitas es $\{\gamma(x,t), K(x,t), \alpha(x,t)\}$, cuyas evoluciones quedan gobernadas por el siguiente sistema de ecuaciones en derivadas parciales:$$\partial_t \gamma = -2\alpha\gamma K$$$$\partial_t K = -\frac{1}{\gamma} \left( \alpha_{xx} - \frac{\gamma_x}{2\gamma}\alpha_x \right) + \frac{1}{3}\alpha K^2$$$$\partial_t\alpha = -\alpha^2 f(\alpha) K$$


Partiendo de las ecuaciones de evolución anteriores, es posible aplicar un cambio de variables estratégico para simplificar la estructura del problema. En primer lugar, introducimos las variables auxiliares $D_\alpha$ y $D_\gamma$, las cuales se definen a partir de las derivadas espaciales de los logaritmos naturales de las funciones originales y de la componente métrica $\gamma$:
\begin{equation}
\gamma := \gamma_{xx}, \quad D_\alpha := \partial_x \ln \alpha \quad \text{and} \quad D_\gamma := \partial_x \ln \gamma.
\end{equation}

Al sustituir estas definiciones y aplicar el álgebra correspondiente, el sistema original se transforma en un nuevo conjunto de ecuaciones para las variables de primer orden:
\begin{align}
\partial_t D_\alpha + \partial_x (\alpha f K) &= 0 \ , \\
\partial_t D_\gamma + \partial_x (2\alpha K) &= 0 \ , \\
\partial_t K + \partial_x (\alpha D_\alpha / \gamma) &= \alpha \left( K^2 - D_\alpha D_\gamma / 2\gamma \right) \ .
\end{align}

A partir de este punto, se observa que la ecuación de evolución para la curvatura externa $K$ puede reescribirse elegantemente como una ley de conservación utilizando la variable escalada $g^{1/2}$:
\begin{equation}
\partial_t \left( \gamma^{1/2} K \right) + \partial_x \left( \alpha D_\alpha / \gamma^{1/2} \right) = 0 \ .
\end{equation}

Si ahora agrupamos nuestras variables en un vector de estado definido como $\vec{v} := (D_\alpha, D_\gamma, \tilde{K})$, donde se introduce la cantidad $\tilde{K} := g^{1/2} K$, las ecuaciones de evolución para las variables de primer orden pueden acoplarse y expresarse como un sistema conservativo de la forma:
\begin{equation}
\partial_t \vec{v} + \partial_x (\mathbf{M} \vec{v}) = 0 \ ,
\end{equation}
donde la matriz característica $\mathbf{M}$ queda completamente determinada por los coeficientes del sistema y viene dada por:
\begin{equation}
\mathbf{M} = \begin{pmatrix}
0 & 0 & \alpha f / g^{1/2} \\
0 & 0 & 2 \alpha / g^{1/2} \\
\alpha / g^{1/2} & 0 & 0
\end{pmatrix} \ .
\end{equation}

Esta sistema matricial es el que vamos a resolver

inicialmente supongo una metrica de minkowski y supongo una pequeña pertubarcion modelado por h.

\begin{align*}
\gamma &= 1 - h'^2 \implies D_\gamma = -2h'h'' / \gamma \ , \\
K &:= K^x_x = -h'' / \gamma^{3/2} \implies \tilde{K} = -h'' / \gamma \ .
\end{align*}

A la funcion h se le dara un perfil gaussiano
\begin{equation}
h = A \exp \left[ (x - x_c)^2 / \sigma^2 \right] \ .
\end{equation}

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp, cumulative_simpson
```

```python
# PARÁMETROS FÍSICOS Y NUMÉRICOS
A = 5.
sigma = 10.
L = 300.          # Tamaño del dominio espacial
x0 = L / 2.       # Centro del pulso gaussiano (150)
f = 1.            # Parámetro de Bona-Massó (f=1: armónico, f=2: 1+log)
tfinal = 100.
N = 2500

# MALLA ESPACIAL ----------------------------------------------------------------
x_numerico = np.linspace(0, L, N, endpoint=False)
dx = x_numerico[1] - x_numerico[0]

# Condiciones iniciales
x = sp.symbols('x', real=True)
h_sym = A * sp.exp(-((x - x0)**2) / (sigma**2))
alpha_sym = sp.Integer(1)

dh_dx_sym = sp.diff(h_sym, x)
d2h_dx2_sym = sp.diff(dh_dx_sym, x)
dalpha_dx_sym = sp.diff(alpha_sym, x)

def evaluar_en_malla(expresion_simbolica, variable_simbolica, malla):
    """
    Evalúa una expresión de SymPy en un array de NumPy.
    Si la expresión es una constante, garantiza que devuelva un array del tamaño de la malla.
    """
    fn = sp.lambdify(variable_simbolica, expresion_simbolica, 'numpy')
    resultado = fn(malla)
    return np.broadcast_to(np.asarray(resultado, dtype=float), malla.shape).copy()

# Evaluaciones en la malla
h_init = evaluar_en_malla(h_sym, x, x_numerico)
dh_dx_evaluada = evaluar_en_malla(dh_dx_sym, x, x_numerico)
d2h_dx2_evaluada = evaluar_en_malla(d2h_dx2_sym, x, x_numerico)
alpha_init = evaluar_en_malla(alpha_sym, x, x_numerico)
dalpha_dx_evaluada = evaluar_en_malla(dalpha_dx_sym, x, x_numerico)

# Condiciones inciales
g_init = 1.0 - dh_dx_evaluada**2

if np.any(g_init <= 0.0):
    raise ValueError(
        f"¡Corte inicial no espacial! max|dh/dx| = {np.max(np.abs(dh_dx_evaluada)):.3f} >= 1. "
        "Reduce la amplitud A o aumenta sigma."
    )

Dg_init = -2.0 * dh_dx_evaluada * d2h_dx2_evaluada / g_init
k_init = -d2h_dx2_evaluada / g_init              # Variable K_tilde = sqrt(g) * K
Dalpha_init = dalpha_dx_evaluada / alpha_init    # Derivada logarítmica: d_x(ln alpha)

# Funciones
def KO(u, dx, eps=0.2):
    """
    Disipación de Kreiss-Oliger de 4to orden vectorizada con condiciones periódicas.
    Nota: eps alrededor de 0.1 - 0.3 es ideal para estabilidad sin sobreamortiguar.
    """
    u_m2 = np.roll(u, 2)
    u_m1 = np.roll(u, 1)
    u_p1 = np.roll(u, -1)
    u_p2 = np.roll(u, -2)

    return -eps * (u_m2 - 4.0*u_m1 + 6.0*u - 4.0*u_p1 + u_p2) / (16.0 * dx)


def reconstruir_desde_derivada_log(D, dx, initial):
    """
    Resuelve dy/dx = D(x) * y(x) usando la solución exacta y(x) = y0 * exp(int D(x) dx)
    con cuadratura de Simpson de 4to orden (vectorizado y preserva positividad).
    """
    integral_D = cumulative_simpson(D, dx=dx, initial=0.0)
    return initial * np.exp(integral_D)


def integrar_RK4_4to_orden(D, dx, initial):
    """
    Si prefieres mantener RK4 explícito, aquí usamos interpolación cúbica centrada
    para D_{i+1/2}, preservando el 4to orden real de convergencia.
    """
    N = len(D)
    y = np.zeros(N)
    y[0] = initial

    # Interpolación de 4to orden para los puntos medios D_{i+1/2}
    D_mid = (-np.roll(D, 1) + 9.0*D + 9.0*np.roll(D, -1) - np.roll(D, -2)) / 16.0

    for i in range(N - 1):
        D0 = D[i]
        D1 = D_mid[i]
        D2 = D1
        D3 = D[i+1]

        k1 = dx * D0 * y[i]
        k2 = dx * D1 * (y[i] + 0.5 * k1)
        k3 = dx * D2 * (y[i] + 0.5 * k2)
        k4 = dx * D3 * (y[i] + k3)

        y[i+1] = y[i] + (k1 + 2.0*k2 + 2.0*k3 + k4) / 6.0

    return y
```

```python
def derivada_espacial_4to_orden(F, dx, periodica=False):
    """
    Calcula dF/dx de 4to orden para un arreglo F de dimensión (n_vars, N).
    """
    if periodica:
        return (
            -np.roll(F, -2, axis=1)
            + 8.0 * np.roll(F, -1, axis=1)
            - 8.0 * np.roll(F, 1, axis=1)
            + np.roll(F, 2, axis=1)
        ) / (12.0 * dx)

    dF_dx = np.zeros_like(F)

    # Interior: centrado de 4to orden
    dF_dx[:, 2:-2] = (
        -F[:, 4:] + 8.0*F[:, 3:-1] - 8.0*F[:, 1:-3] + F[:, :-4]
    ) / (12.0 * dx)

    # Fronteras izquierdas: descentrado de 4to orden
    dF_dx[:, 0] = (
        -25.0*F[:, 0] + 48.0*F[:, 1] - 36.0*F[:, 2] + 16.0*F[:, 3] - 3.0*F[:, 4]
    ) / (12.0 * dx)
    dF_dx[:, 1] = (
        -3.0*F[:, 0] - 10.0*F[:, 1] + 18.0*F[:, 2] - 6.0*F[:, 3] + F[:, 4]
    ) / (12.0 * dx)

    # Fronteras derechas: descentrado de 4to orden
    dF_dx[:, -2] = (
        3.0*F[:, -1] + 10.0*F[:, -2] - 18.0*F[:, -3] + 6.0*F[:, -4] - F[:, -5]
    ) / (12.0 * dx)
    dF_dx[:, -1] = (
        25.0*F[:, -1] - 48.0*F[:, -2] + 36.0*F[:, -3] - 16.0*F[:, -4] + 3.0*F[:, -5]
    ) / (12.0 * dx)

    return dF_dx


def RHS(t, y_global, N, dx, f):
    Dalpha, Dg, k = y_global.reshape((3, N))

    # Reconstrucción espacial de 4to orden vectorizada (sin bucles en Python)
    alpha = reconstruir_desde_derivada_log(Dalpha, dx=dx, initial=1.0)
    g     = reconstruir_desde_derivada_log(Dg, dx=dx, initial=1.0)

    inv_sqrt_g = 1.0 / np.sqrt(g)
    vel_escala = alpha * inv_sqrt_g

   # Flujos conservativos F(u)
    F = np.empty((3, N))
    F[0, :] = f * vel_escala * k
    F[1, :] = 2.0 * vel_escala * k
    F[2, :] = vel_escala * Dalpha

    # Derivada espacial del flujo
    dF_dx = derivada_espacial_4to_orden(F, dx, periodica=True)

    # Sistema evolutivo
    dDalpha_dt = -dF_dx[0] + KO(Dalpha, dx, eps=0.2)
    dDg_dt     = -dF_dx[1] + KO(Dg, dx, eps=0.2)
    dk_dt      = -dF_dx[2] + KO(k, dx, eps=0.2)

    return np.concatenate([dDalpha_dt, dDg_dt, dk_dt])


```

```python
# --- INTEGRACIÓN TEMPORAL ----------------------------------------
y_inicial = np.concatenate([Dalpha_init, Dg_init, k_init])

t_span = (0.0, tfinal)
t_eval = np.linspace(0.0, tfinal, 1250)

# Paso máximo tipo CFL: dt_max <= 0.5 * dx / sqrt(f)
dt_cfl = 0.5 * dx / np.sqrt(max(f, 1.0))
solucion = solve_ivp(
    RHS,
    t_span,
    y_inicial,
    t_eval=t_eval,
    args=(N, dx, f),
    method='RK45',
    rtol=1e-8,
    atol=1e-10,
    max_step=dt_cfl
)
```

```python
# Índices de 4 instantes de tiempo equiespaciados
num_pasos = solucion.y.shape[1]
time_indices = np.linspace(0, num_pasos - 1, 4, dtype=int)

# Creamos 3 paneles verticales que comparten el mismo eje x
fig, axes = plt.subplots(3, 1, figsize=(10, 10), sharex=True)

for j in time_indices:
    t_actual = solucion.t[j]

    # Extraemos las 3 variables de estado en el instante j
    Dalpha_j, Dg_j, k_j = solucion.y[:, j].reshape((3, N))

    # Reconstruimos alpha y g (usa la misma función que tengas en RHS)
    alpha_j = reconstruir_desde_derivada_log(Dalpha_j, dx=dx, initial=1.0)
    g_j     = reconstruir_desde_derivada_log(Dg_j, dx=dx, initial=1.0)

    etiqueta = f't = {t_actual:.2f}'
    axes[0].plot(x_numerico, k_j, label=etiqueta)
    axes[1].plot(x_numerico, alpha_j, label=etiqueta)
    axes[2].plot(x_numerico, g_j, label=etiqueta)

# Configuración de etiquetas y títulos
axes[0].set_title('Evolución Temporal del Sistema 1+1 (Bona-Massó, $f = ' + f'{f}$)')
axes[0].set_ylabel(r'Curvatura $\tilde{K}$')
axes[1].set_ylabel(r'Lapso $\alpha$')
axes[2].set_ylabel(r'Métrica $\gamma_{xx}$')
axes[2].set_xlabel('$x$')

for ax in axes:
    ax.grid(True, alpha=0.4)
    ax.legend(bbox_to_anchor=(1.02, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

La solución obtenida corresponde a una matriz bidimensional. En ella, las filas contienen las variables del sistema: desde la fila 1 hasta la N se encuentran los valores de $D_α$	, desde la fila N hasta la 2N los de $D_g$
, y desde la fila 2N hasta la 3N los de K. Las columnas representan los distintos instantes temporales considerados, en este caso 50, distribuidos en el intervalo de 0 a 100.
