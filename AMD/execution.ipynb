```python
from AMD import AMD
import sympy as sp
from IPython.display import display
```

```python
dt, dr, dtheta, dphi = sp.symbols(
    r'dt dr d\theta d\phi'
)

# Coordenadas
t, r, theta, phi, M = sp.symbols(
    r't r \theta \phi M',
    positive=True
)
diffs0 = [dt, dr, dtheta, dphi]
variables0 = [t, r, theta, phi]
# Factor
f = 1 - 2*M/r

# ds^2
ds2 = (
    -f*dt**2
    + dr**2/f
    + r**2*dtheta**2
    + r**2*sp.sin(theta)**2*dphi**2
)
display(ds2)
```

```python
miespacio1 = AMD(ds2,diffs0,variables0)

display(miespacio1.christophel3D())
display(miespacio1.shift())
display(miespacio1.extrinsicCurvature())
```

```python
# Coordenadas
t, r, theta, phi = sp.symbols(r't r \theta \phi', real=True)

# Diferenciales
dt, dr, dtheta, dphi = sp.symbols(r'dt dr d\theta d\phi')

# Parámetro
M = sp.symbols('M', positive=True)

diffs2 = [dt, dr, dtheta, dphi]
variables2 = [t, r, theta, phi]
# Elemento de línea
ds3 = (
    -dt**2
    + (dr + sp.sqrt(2*M/r)*dt)**2
    + r**2*(dtheta**2 + sp.sin(theta)**2 * dphi**2)
)

sp.expand(ds3)
```

```python
miespacio2 = AMD(ds3,diffs2,variables2)

display(miespacio2.christophel3D())
display(miespacio2.extrinsicCurvature())
```
